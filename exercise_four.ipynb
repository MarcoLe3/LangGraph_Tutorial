```python
from typing import TypedDict
from langgraph.graph import StateGraph, START, END
```

```python
class AgentState(TypedDict):
  num_one: int
  num_two: int
  operation_one: str
  num_three: int
  num_four: int
  operation_two: str
  final_number_one: int
  final_number_two: int

def add_one(state: AgentState) -> AgentState:
  state["final_number_one"] = state["num_one"] + state["num_two"]
  return state

def add_two(state: AgentState) -> AgentState:
  state["final_number_two"] = state["num_three"] + state["num_four"]
  return state

def subtract_one(state: AgentState) -> AgentState:
  state["final_number_one"] = state["num_one"] - state["num_two"]
  return state

def subtract_two(state: AgentState) -> AgentState:
  state["final_number_two"] = state["num_three"] - state["num_four"]
  return state

def decide_one(state: AgentState) -> AgentState:
  if state["operation_one"] == "+":
    return "add_operation_one"
  else:
    return "subtract_operation_one"

def decide_two(state: AgentState) -> AgentState:
  if state["operation_two"] == "+":
    return "add_operation_two"
  else:
    return "subtract_operation_two"
```

```python
graph = StateGraph(AgentState)
graph.add_node("add_one", add_one)
graph.add_node("add_two", add_two)
graph.add_node("subtract_one", subtract_one)
graph.add_node("subtract_two", subtract_two)
graph.add_node("router_one", lambda state:state)
graph.add_node("router_two", lambda state:state)

graph.add_edge(START, "router_one")

graph.add_conditional_edges(
    "router_one",
    decide_one,
    {
        "add_operation_one": "add_one",
        "subtract_operation_one": "subtract_one"
    }
)

graph.add_edge("add_one", "router_two")
graph.add_edge("subtract_one", "router_two")

graph.add_conditional_edges(
    "router_two",
    decide_two,
    {
        "add_operation_two": "add_two",
        "subtract_operation_two": "subtract_two"
    }
)

graph.add_edge("add_two", END)
graph.add_edge("subtract_two", END)

app = graph.compile()
```

```python
app.invoke({
  "num_one": 1,
  "num_two": 2,
  "operation_one": "+",
  "num_three": 3,
  "num_four": 4,
  "operation_two": "-"
})
```
