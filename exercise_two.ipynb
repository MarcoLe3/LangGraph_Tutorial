```python
from typing import TypedDict, List
from langgraph.graph import StateGraph
```

```python
class AgentState(TypedDict):
  values: List[int]
  name: str
  operation: str
  result: str
```

```python
import math

def process_operatoration(state: AgentState) -> AgentState:
  """This function handles multiple values"""
  if state["operation"] == "+":
    state["result"] = f"Hi there {state['name']}, your sum is {sum(state['values'])}"
  else:
    state["result"] = f"Hi there {state["name"]}, your multiply is {math.prod(state["values"])}"
  return state
```

```python
graph = StateGraph(AgentState)
graph.add_node("operator", process_operatoration)
graph.set_entry_point("operator")
graph.set_finish_point("operator")
app = graph.compile()
```

```python
app.invoke({"values": [1,2,3,4], "name": "John", "operation": "+"})
```
